```python
# prompt: Suppose you want to test whether the maths score of students who have not taken the test preparation course is less than the maths score of students who have taken any test preparation course. What will be the conclusion based on the p-value?
# Note - Assume that the population standard deviations are equal.

# Sample data for students who have not taken the test preparation course
not_taken_course_scores = [61.00, 64.00, 67.00, 73.67]

# Sample data for students who have taken any test preparation course
taken_course_scores = [77.00, 89.33, 85.00, 90.00, 92.00]

# Set the null hypothesis mean
null_hypothesis_mean = np.mean(taken_course_scores)

# Perform two-sample t-test with equal variances
t_statistic, p_value = stats.ttest_ind(not_taken_course_scores, taken_course_scores, equal_var=True, alternative='less')

# Set the significance level
alpha = 0.05

# Interpret the result based on the p-value
if p_value < alpha:
    conclusion = "As the p-value is less than 0.05, you do have enough evidence to say that the maths score of students who have not taken the test preparation course is less than the maths score of students who have taken any test preparation course."
else:
    conclusion = "As the p-value is greater than or equal to 0.05, you do not have enough evidence to say that the maths score of students who have not taken the test preparation course is less than the maths score of students who have taken any test preparation course."

print(conclusion)


As the p-value is less than 0.05, you do have enough evidence to say that the maths score of students who have not taken the test preparation course is less than the maths score of students who have taken any test preparation course.
```

```python
# prompt: Suppose you want to test that whether there is any difference in means of student performance in reading, writing, and maths. What will be the conclusion based on the p-value?

import pandas as pd
import numpy as np
from scipy import stats

# Load data
df = pd.read_csv('/content/drive/MyDrive/GL_AIML/Statistics/HandsOnQuiz/students.csv')

# Extract relevant columns
reading_scores = df['reading_score'].dropna().values
writing_scores = df['writing_score'].dropna().values
maths_scores = df['maths_score'].dropna().values

# Perform one-way ANOVA
f_statistic, p_value = stats.f_oneway(reading_scores, writing_scores, maths_scores)

# Set the significance level
alpha = 0.05

# Interpret the result based on the p-value
if p_value < alpha:
    conclusion = "As the p-value is less than 0.05, you do have enough evidence to say that there is a difference in means of student performance in reading, writing, and maths."
else:
    conclusion = "As the p-value is greater than or equal to 0.05, you do not have enough evidence to say that there is a difference in means of student performance in reading, writing, and maths."

print(conclusion)
```

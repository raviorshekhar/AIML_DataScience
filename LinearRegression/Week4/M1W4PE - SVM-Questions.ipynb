# 8. Do train and test split with stratify sampling on Outcome variable to maintain the distribution of dependent variable

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.loc[:,df.columns != 'Outcome' ], df['Outcome'], stratify = df['Outcome'], random_state=11)
```

```python
X_train.shape

```

# 9. Train Support Vector Machine Model

```python
from sklearn.svm import SVC
svc = SVC()
svc.fit(X_train, y_train)
training_score = svc.score(X_train, y_train)
print("Training score: ", training_score)
test_score = svc.score(X_test, y_test)
print("Test score: ", test_score)
```

```python
print("Accuracy on training set: {:.2f}".format(svc.score(X_train, y_train)))
print("Accuracy on test set: {:.2f}".format(svc.score(X_test, y_test)))
```

```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(X_train, y_train)

print("Accuracy on training set: {:.2f}".format(svc.score(X_train, y_train)))
print("Accuracy on test set: {:.2f}".format(svc.score(X_test, y_test)))
```

# 10. Scale the data points using MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# 11. Fit SVM Model on Scale data and give your observation

```python
svc = SVC()
svc.fit(X_train_scaled, y_train)

print("Accuracy on training set: {:.2f}".format(svc.score(X_train_scaled, y_train)))
print("Accuracy on test set: {:.2f}".format(svc.score(X_test_scaled, y_test)))
```

# 12. Try improving the model accuracy using C=1000

```python
svc = SVC(C=1000)
svc.fit(X_train_scaled, y_train)

print("Accuracy on training set: {:.3f}".format(
    svc.score(X_train_scaled, y_train)))
print("Accuracy on test set: {:.3f}".format(svc.score(X_test_scaled, y_test)))
```

# Here, increasing C allows us to improve the model, resulting in 95% train set accuracy.
